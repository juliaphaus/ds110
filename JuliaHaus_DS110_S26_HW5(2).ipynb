# Problem 1: Regular Expressions (20 pts [10,6,4])

Each of these sub-problems requires the use of regular expressions.

a, 10 pts) A recipe website uses the following URL format for its recipes "https://awesomerecipes.com/recipes/abcdefgh-12/recipe-name". Users who post their recipes have unique ids (e.g., "abcdefgh-12") consisting of 8 letters followed by a hyphen and 2 numbers. The recipe name is either a single word or has a single hyphen separating the title, for example, "donuts" or "creme-donuts".

Write a function `extract_user_and_recipe(text)` that returns the unique ID and the recipe name from the URL. Ensure your function can extract all the words in the hyphen separated title.

```python
import re
def extract_user_and_recipe(text):
  pattern = r"/recipes/([a-z]{8}-\d{2})/([a-z-]+)/?$"
  match = re.search(pattern, text)
  if match:
    user_id = match.group(1)
    recipe_name = match.group(2)
    return f"{user_id}, {recipe_name}"
  return "No matches"
```

```python
# Expect 'abcxyzwv-34, donuts':
text1 = 'https://awesomerecipes.com/recipes/abcxyzwv-34/donuts'
# Expect 'abcxyzwv-34, creme-donuts':
text2 = 'https://awesomerecipes.com/recipes/abcxyzwv-34/creme-donuts'

print(extract_user_and_recipe(text1))
print(extract_user_and_recipe(text2))
```

b, 6 pts) A college's IT department has decided to update the unique IDs of their internal database of students and faculty/staff by changing to a new format. The old format consisted of 3 letter FAC for faculty, STA for staff, and STU for student, followed by 9 digits. The new format consists of a single letter, U for faculty/staff and u for students, followed by the original 9 digits. Write a function `convert_id(text)` which accepts as input a university ID and returns the re-formatted ID.

```python
def convert_id(text):
  if re.search(r"^(FAC|STA)(\d{9})", text):
    return re.sub(r"(FAC|STA)(\d{9})", r"U\2", text)
  elif re.search(r"^STU", text):
    return re.sub(r"^STU(\d{9})", r"u\1", text)
  return text
```

```python
print(convert_id("FAC123456789"))  # Expect 'U123456789'
print(convert_id("STA123456789"))  # Expect 'U123456789'
print(convert_id("STU987654321"))  # Expect 'u987654321'
```

c, 4 pts) An API (application programming interface) endpoint is a specific URL that allows clients (like web browsers, mobile apps, or other servers) to access specific resources or perform certain actions. API endpoints often have query parameters, for example /data?XML or/users?id=123. Write a function `extract_query(text)` which can extract queries that follow a '?' consisting of either 1 or more alphanumeric characters or 1 or more alphanumeric characters that are followed by an equal sign and a string of 1 or more alphanumeric characters. The return value should be a string.

```python
def extract_query(text):
  pattern_1 = r"\?([a-zA-Z0-9=]+)"
  match = re.search(pattern_1, text)
  if match:
    return match.group(1)
  return ""
```

```python
print(extract_query("/data?XML"))           # Expect 'XML'
print(extract_query("/users?id=123"))       # Expect 'id=123'
print(extract_query("/search?query=books")) # Output: 'query=books'
```

# Problem 4: The radar image class (18 pts [2, 4, 2, 5, 5])

You work for a ship-motion prediction company that needs to create a Python class in order to accumulate continuously streaming radar images. Each radar image is a 2D NumPy array of dimension 10 x 10. The class should be named `RadarImages`. You need to design this class to do the following:

a, 2 pts) Initialize the class with an attribute called `img_list` that is an empty list and an attribute called `max_img_list_size` that is set from an integer `max_img_list_size` passed as an input argument in the class constructor.

b, 4 pts) Implement an `add_radar_image(matrix)` member function which appends a 10x10 2D numpy array to `image_list`. If you call this function and the length of the list is `max_img_list_size` (before appending), remove the 1st element of list and append the new image to the end of the list.

c, 2 pts) Implement a `flush_images()` member function which sets the `img_list` attribute to an empty list.

d, 5 pts) Implement an `average_radar_images()` member function which returns an average image matrix. Specifically, the average image matrix is the sum of all the images in the list divided by the total number of images. If the image list is empty, return a 10x10 matrix of zeros.

e, 5 pts) Implement a `blur_radar_image(matrix, index)` function which takes a 10x10 2D NumPy array and an integer index as input arguments. The function should add the input argument `matrix` to the radar image at position index in the `img_list`. If the index is out of bounds, raise an `IndexError`. The syntax to raise an `IndexError` is `raise IndexError("Index out of bounds")`.

```python
import numpy as np
class RadarImages:
  def __init__(self, max_img_list_size):
    self.image_list = []
    self.max_img_list_size = max_img_list_size

  def add_radar_images(self, matrix):
    if len(self.image_list) == self.max_img_list_size:
      self.image_list.pop(0)
    self.image_list.append(matrix)

  def flush_images(self):
    self.image_list = []

  def average_radar_image(self):
    if not self.image_list:
      return np.zeros((10, 10))
    total_sum = sum(self.image_list)
    return total_sum / len(self.image_list)

  def blur_radar_image(self, matrix, index):
    try:
      self.image_list[index] = self.image_list[index] + matrix
    except IndexError:
      raise IndexError("Index out of bounds")
```

```python
import numpy as np

radar_images = RadarImages(5)
rng = np.random.default_rng(42)
image1 = rng.random((10, 10))
image2 = rng.random((10, 10))
image3 = rng.random((10, 10))
image4 = rng.random((10, 10))
image5 = rng.random((10, 10))
image6 = rng.random((10, 10))

# TEST 1 should print 10 x 10 matrix of zeros and returns True if the returned array is equal to matrix of zeros
print(radar_images.average_radar_image())
print(np.array_equal(np.zeros((10, 10)), radar_images.average_radar_image()))
```
