# Tool focus:  The correlation heatmap

A nice trick for making your correlation results look nicer is to create a Seaborn heatmap with them.  Seaborn is a module for visualization that we'll cover later; this is just one example of what it can do.

```python
import seaborn as sns

df = pd.read_csv('cereal.csv',index_col="name")

sns.heatmap(df.corr(numeric_only=True))
```

Cool-looking, but a little hard to interpret.  We can set the optional argument *annot* to True to get the actual correlation numbers back.  The numbers are going to look a little cramped unless we also increase the size of the figure, so we'll do that, too.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(16, 6)) # Increase figure size to width=16, height=6

sns.heatmap(df.corr(numeric_only=True), annot=True)
```
